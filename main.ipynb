## Task 7: Data Cleaning

```python
import re

clean_df = df.copy()

# 1. Properly convert PascalCase to snake_case using Regex
# (e.g., 'BusinessTravel' -> 'business_travel', 'MonthlyIncome' -> 'monthly_income')
clean_df.columns = [re.sub(r'(?<!^)(?=[A-Z])', '_', col).lower() for col in clean_df.columns]

# 2. Check & drop duplicate rows
clean_df = clean_df.drop_duplicates()

# 3. Drop uninformative constant columns
clean_df.drop(columns=['employee_count', 'over_18', 'standard_hours'], inplace=True, errors='ignore')

# 4. Convert categorical text columns to 'category' dtype
cat_cols = [
    'attrition', 
    'business_travel', 
    'department', 
    'education_field', 
    'gender', 
    'job_role', 
    'marital_status', 
    'over_time'
]

for col in cat_cols:
    if col in clean_df.columns:
        clean_df[col] = clean_df[col].astype('category')

print(f"Cleaned Dataset Shape: {clean_df.shape}")
print("\nCleaned Column Names:")
print(clean_df.columns.tolist()[:10])
```

## Task 8: Data Analysis (3 Questions & Explanations)

```python
# Cell 7: Question 1 - Department Attrition Rate Analysis
dept_attrition = clean_df.groupby('department')['attrition'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index()
dept_attrition.rename(columns={'attrition': 'attrition_rate_pct'}, inplace=True)
display(dept_attrition)
```

**Insight Q1:** Sales displays the highest attrition rate relative to Research & Development and HR. Commission structures and travel requirements often drive higher turnover in sales units.

```python
# Cell 8: Question 2 - OverTime Impact on Monthly Income and Attrition
overtime_analysis = clean_df.groupby(['over_time', 'attrition'])['monthly_income'].agg(['mean', 'count']).reset_index()
display(overtime_analysis)
```

**Insight Q2:** Employees working overtime exhibit a significantly higher rate of departure despite comparable income tiers, indicating burnout factors beyond monetary compensation.

```python
# Cell 9: Question 3 - Average Tenure and Income by Job Role
role_summary = clean_df.groupby('job_role').agg(
    avg_income=('monthly_income', 'mean'),
    avg_years=('years_at_company', 'mean'),
    employee_count=('age', 'count')
).sort_values(by='avg_income', ascending=False).reset_index()
display(role_summary)
```

**Insight Q3:** Managers and Research Directors maintain the highest stability (tenure) and salary bands, while Lab Technicians and Sales Representatives show lower retention horizons.

## Task 9: Multi-Sheet Excel Export (Project_Report.xlsx)

```python
# Cell 10: Formatted Excel File Generation
import openpyxl

file_path = "reports/Project_Report.xlsx"

# Create Excel writer object
with pd.ExcelWriter(file_path, engine='openpyxl') as writer:
    # Sheet 1: Cleaned Data
    clean_df.to_excel(writer, sheet_name='Cleaned Data', index=False)
    
    # Sheet 2: Analysis Results
    role_summary.to_excel(writer, sheet_name='Analysis', index=False, startrow=0)
    dept_attrition.to_excel(writer, sheet_name='Analysis', index=False, startrow=len(role_summary) + 4)
    
    # Sheet 3: Summary Sheet
    summary_data = pd.DataFrame({
        'Metric': ['Total Records Evaluated', 'Average Monthly Income', 'Overall Attrition Rate (%)', 'Highest Risk Department'],
        'Value': [len(clean_df), f"${clean_df['monthly_income'].mean():,.2f}", f"{(clean_df['attrition'] == 'Yes').mean()*100:.2f}%", 'Sales']
    })
    summary_data.to_excel(writer, sheet_name='Summary', index=False, startrow=2)

# Format Excel appearance using openpyxl
wb = openpyxl.load_workbook(file_path)

# Apply formatting (bold headers, column width adjustments)
for ws in wb.worksheets:
    for col in ws.columns:
        max_len = max(len(str(cell.value or '')) for cell in col)
        col_letter = openpyxl.utils.get_column_letter(col[0].column)
        ws.column_dimensions[col_letter].width = max(max_len + 3, 12)
        
    for cell in ws[1]:
        cell.font = openpyxl.styles.Font(bold=True)

wb.save(file_path)
print(f"Successfully generated formatted Excel report at {file_path}")
```
